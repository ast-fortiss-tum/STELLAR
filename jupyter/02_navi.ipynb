# STELLAR Navigation

We will run experiments for navigational use case in this notebooks. The evaluated systems work as following:

#### Input

System takes textual user request as the input.

Input example:

```text
Could you guide to a cafe with American food, rated 4 stars, and contactless payment? Parking should be available.
```

#### Output
System outputs textual response and list of offered navigational recomendations. Each recommendation is a point of interest (POI), with provided structured information about the POI.

Output example:

```text
Based on your preference for American food, a 4-star rating, contactless payment, and available parking, I recommend these options.
```

POI example:

```json
{
    "title": "The Living Room Cafe",
    "categories": [
        "Food",
        "Breakfast & Brunch",
        "Cafes",
        "Restaurants",
        "Coffee & Tea",
        "American (Traditional)"
    ],
    "address": "701 S 5th St",
    "location": {
        "lat": 39.9402868,
        "lng": -75.150867
    },
    "business_hours_status": null,
    "payment_methods": null,
    "rating": 4.5,
    "price_range": "$$",
    "fuel_prices": null,
    "fuel_types": null,
    "gas_station_brand": null,
    "restaurant_brand": null,
    "food_types": [],
    "parking": "available",
    "charging": null,
    "availability": null
}
```

#### Evaluation

We evaluate system responses with two metrics:

- **Content Fitness:** we measure if the provided POI correspond to the user's request based on the values of the filed of the structured POI output.
- **Answer Fitness:** we evaluate if the textual output follows the linguistic requirements. In particular, we judge if the answer is request-oriented, proactive, and clear.

## Configuring

The provided script `run_tests_navi.py` has the following parameters:

| Argument | Type | Default | Description |
|----------|------|---------|-------------|
| `--sut` | str | `IPA_LOS` | Select the System Under Test (`IPA_LOS`, `IPA_YELP`) |
| `--population_size` | int | `2` | Population size for GA |
| `--n_generations` | int | `2` | Number of generations |
| `--seed` | int | `4` | Random seed |
| `--max_time` | str | `None` | Maximal execution time as `hh:mm:ss` |
| `--th_answer` | float | `0.75` | Threshold for answer fitness |
| `--th_content` | float | `0.75` | Threshold for content fitness |
| `--algorithm` | str | `nsga2d` | Algorithm (`rs`, `nsga2`, `nsga2d`) |
| `--results_folder` | str | `results` | Path to store results |
| `--no_wandb` | flag | - | Turn off W&B logging |
| `--use_rag` | flag | - | Use RAG in test generation |
| `--judge` | str | `gpt-4o-mini` | LLM used for judgement |
| `--judge_weights` | list | `[1, 0, 0.5]` | Weights for the judge dimensions |
| `--features_config` | str | `configs/features_simple_judge.json` | Path to feature config |

You can modify features by extendending the lists of values in the feature configuration files. You can also remove certain values from the config.
